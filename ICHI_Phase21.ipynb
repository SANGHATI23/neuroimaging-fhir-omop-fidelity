```python
# Source-level inspection: execute ONLY the attested original deterministic function definitions.
# Do NOT run the historical runner on old medication data or execute cells doing file reads / external actions.
want={'nonempty_warning','compute_metrics','split_warning_tokens','join_warning_tokens','warning_count'}
function_defs={}; assignment_nodes={}; original_status_literals=set(); event_columns=set(); input_read_functions=Counter(); suspected_audit_producers=0
for cell_index,src in code_cells:
    tree=ast.parse(src)
    for node in ast.walk(tree):
        if isinstance(node,ast.FunctionDef) and node.name in want:
            function_defs[node.name]=(cell_index,ast.get_source_segment(src,node))
        if isinstance(node,ast.Assign):
            for target in node.targets:
                if isinstance(target,ast.Name) and target.id in {'LOSS_STATUSES','NON_LINEAGE_ELIGIBLE_STATUSES'}:
                    assignment_nodes[target.id]=ast.get_source_segment(src,node.value)
        if isinstance(node,ast.AnnAssign) and isinstance(node.target,ast.Name) and node.target.id in {'LOSS_STATUSES','NON_LINEAGE_ELIGIBLE_STATUSES'}:
            assignment_nodes[node.target.id]=ast.get_source_segment(src,node.value)
        if isinstance(node,ast.Constant) and isinstance(node.value,str):
            val=node.value.strip()
            if val in {'PRESERVED','UNCHANGED','NORMALIZED','AMBIGUOUS','UNMAPPED','DROPPED','TRACEABILITY_LOSS'}:original_status_literals.add(val)
        if isinstance(node,ast.Subscript) and isinstance(node.slice,ast.Constant) and isinstance(node.slice.value,str):
            if node.slice.value in {'source_df_index','source_resource_type','fidelity_status','warning_code','is_mapping_event','target_omop_table','target_omop_record_id'}:event_columns.add(node.slice.value)
        if isinstance(node,ast.Call) and isinstance(node.func,ast.Attribute):
            if node.func.attr in {'read_csv','read_parquet','read_json'}:input_read_functions[node.func.attr]+=1
assert all(name in function_defs for name in ('nonempty_warning','compute_metrics'))
assert function_defs['compute_metrics'][0]>function_defs['nonempty_warning'][0]-1
# Only unmodified, verified original function source; builtins/globals supplied below.
original_namespace={'np':np,'pd':pd}
for required in ('LOSS_STATUSES','NON_LINEAGE_ELIGIBLE_STATUSES'):
    expr=assignment_nodes.get(required)
    if expr is not None:
        parsed=ast.parse(expr,mode='eval')
        if isinstance(parsed.body,(ast.Constant,ast.Set,ast.List,ast.Tuple,ast.Dict)):
            value=ast.literal_eval(parsed.body)
        elif (isinstance(parsed.body,ast.Call) and isinstance(parsed.body.func,ast.Name)
              and parsed.body.func.id in {'set','frozenset','tuple','list'} and len(parsed.body.args)==1 and not parsed.body.keywords):
            value=getattr(__builtins__,parsed.body.func.id, None) if not isinstance(__builtins__,dict) else __builtins__.get(parsed.body.func.id)
            assert value is not None
            value=value(ast.literal_eval(parsed.body.args[0]))
        else:value=None
        if isinstance(value,(set,frozenset,tuple,list)):
            assert all(isinstance(x,str) for x in value)
            original_namespace[required]=set(value)
# Original notebook final function definition wins (cell 9 replaces earlier cell 7).
for fn in ('nonempty_warning','compute_metrics','split_warning_tokens','join_warning_tokens','warning_count'):
    if fn in function_defs:
        index,src=function_defs[fn]
        exec(compile(ast.parse(src),str(RUNNER)+'::verified_cell_'+str(index),'exec'),original_namespace)
original_compute=original_namespace['compute_metrics']
method_ready=all(k in original_namespace for k in ('LOSS_STATUSES','NON_LINEAGE_ELIGIBLE_STATUSES'))
assert not method_ready or 'UNMAPPED' in original_namespace['NON_LINEAGE_ELIGIBLE_STATUSES'],'Unexpected original denominator taxonomy; manual inspection required.'
print('Original last-defined compute_metrics cell:',function_defs['compute_metrics'][0])
print('Original frozen status constants recovered:',method_ready,'; original status tokens:',sorted(original_status_literals))
```
